```python
# https://scikit-learn.org/stable/modules/generated/sklearn.datasets.load_iris.html
# https://scikit-learn.org/stable/modules/generated/sklearn.datasets.load_wine.html
# https://scikit-learn.org/stable/modules/generated/sklearn.datasets.load_diabetes.html

from sklearn.datasets import load_iris, load_wine, load_diabetes
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.linear_model import SGDClassifier, LinearRegression
import numpy as np
```

```python

iris = load_iris()
wine = load_wine()
diabetes = load_diabetes()
```

```python
X_tr_iris, X_te_iris, y_tr_iris, y_te_iris = train_test_split(iris.data, iris.target, test_size=0.2)
X_tr_wine, X_te_wine, y_tr_wine, y_te_wine = train_test_split(wine.data, wine.target, test_size=0.2)
X_tr_diab, X_te_diab, y_tr_diab, y_te_diab = train_test_split(diabetes.data, diabetes.target, test_size=0.2)
```

```python
model_iris = svm.SVC()
model_iris.fit(X_tr_iris, y_tr_iris)
```

```python
model_wine = SGDClassifier(loss="hinge", penalty="l2", max_iter=100, random_state=42)
model_wine.fit(X_tr_wine, y_tr_wine)
```

```python
model_diabetes = LinearRegression()
model_diabetes.fit(X_tr_diab, y_tr_diab)
```

```python
from sklearn.metrics import accuracy_score, classification_report
from sklearn.metrics import confusion_matrix, f1_score, accuracy_score, mean_squared_error, root_mean_squared_error

previsoes_iris = model_iris.predict(X_te_iris)

acuracia_iris = accuracy_score(y_te_iris, previsoes_iris)
print(acuracia_iris)

print(classification_report(y_te_iris, previsoes_iris))

```

```python
cm = confusion_matrix(y_te_iris, previsoes_iris)
print(cm)
```

```python
from sklearn.metrics import mean_squared_error, r2_score

previsoes_diabetes = model_diabetes.predict(X_te_diab)

mse = mean_squared_error(y_te_diab, previsoes_diabetes)
r2 = r2_score(y_te_diab, previsoes_diabetes)

print(mse)
print(r2)
```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.linear_model import LogisticRegression, SGDClassifier
from sklearn import svm
from sklearn.metrics import accuracy_score
import numpy as np
```

```python
X = iris.data
y = iris.target

# 1. Usar os dados do Iris que já carregou no notebook (x e y)
# 2. Configurar o StratifiedKFold para 5 divisões (folds)
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

modelos = {
    "SVC": svm.SVC(),
    "SGD Classifier": SGDClassifier(loss="hinge", penalty="l2", max_iter=100, random_state=42),
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42)
}

for nome, modelo in modelos.items():
    lista_acuracias = []
    lista_f1 = []
    
    for treino_idx, teste_idx in skf.split(X, y):
        X_treino_fold, X_teste_fold = X[treino_idx], X[teste_idx]
        y_treino_fold, y_teste_fold = y[treino_idx], y[teste_idx]
        
        # Treinar o modelo
        modelo.fit(X_treino_fold, y_treino_fold)
        
        # Prever e calcular a acurácia
        previsoes = modelo.predict(X_teste_fold)
        acuracia = accuracy_score(y_teste_fold, previsoes)
        lista_acuracias.append(acuracia)
        f1 = f1_score(y_teste_fold, previsoes, average='macro')
        lista_f1.append(f1)

    acuracia_media = np.mean(lista_acuracias)
    f1_medio = np.mean(lista_f1)
    print(f"Modelo: {nome} | Acurácia Média: {acuracia_media * 100:.2f}% | F1-Score Médio: {f1_medio * 100:.2f}%")
```
